```python
# DEMO A1 — memuat data
# Load the data by skipping initial header rows and providing explicit column names
column_names = [
    'NO',
    'TAHUN',
    'TEMPAT_WISATA',
    'WISATAWAN_LOKAL',
    'WISATAWAN_NUSANTARA',
    'WISATAWAN_MANCA_NEGARA',
    'JUMLAH_TOTAL'
]
data = pd.read_csv(FILE, skiprows=3, names=column_names)
data.head()
```
