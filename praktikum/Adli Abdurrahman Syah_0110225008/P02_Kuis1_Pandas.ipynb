**Data:** `kunjungan_perpustakaan_2025.csv` — catatan kunjungan perpustakaan kampus, Februari–Juni 2025.
Satu baris = satu kunjungan.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('kunjungan_perpustakaan_2025.csv')
df.head()
```

---
### Tugas 1 (15 poin) — Memuat dan memeriksa

```python
# a. Cetak jumlah baris dan kolom
print("Jumlah baris dan kolom:", df.shape)

# b. Cetak jumlah nilai kosong (missing) untuk setiap kolom
print("\nJumlah nilai kosong tiap kolom:\n", df.isnull().sum())

# c. Cetak berapa mahasiswa unik (nim_hash) yang pernah berkunjung
print("\nJumlah mahasiswa unik:", df['nim_hash'].nunique())
```

---
### Tugas 2 (20 poin) — Filter

```python
# Ambil kunjungan mahasiswa prodi 'Teknik Informatika' dengan durasi_menit > 120.
df_filtered = df[(df['prodi'] == 'Teknik Informatika') & (df['durasi_menit'] > 120)]

# Cetak jumlah barisnya dan rata-rata jumlah_pinjam pada kelompok tersebut (2 desimal).
jumlah_baris = df_filtered.shape[0]
rata_pinjam = df_filtered['jumlah_pinjam'].mean()

print(f"Jumlah baris: {jumlah_baris}")
print(f"Rata-rata jumlah pinjam: {rata_pinjam:.2f}")
```

---
### Tugas 3 (20 poin) — Kolom turunan

```python
# a. Buat kolom 'durasi_jam' = durasi_menit / 60, dibulatkan 2 desimal
df['durasi_jam'] = (df['durasi_menit'] / 60).round(2)

# b. Buat kolom 'lama' berisi 'Panjang' jika durasi_menit >= 90, selain itu 'Singkat'
df['lama'] = np.where(df['durasi_menit'] >= 90, 'Panjang', 'Singkat')

# c. Tampilkan 5 kunjungan dengan durasi terlama (kolom: prodi, jenis_ruang, durasi_jam)
lima_terlama = df.nlargest(5, 'durasi_menit')[['prodi', 'jenis_ruang', 'durasi_jam']]
lima_terlama
```

---
### Tugas 4 (25 poin) — Agregasi

```python
# Buat tabel ringkasan per 'jenis_ruang'
ringkas = df.groupby('jenis_ruang').agg(
    jumlah_kunjungan=('jenis_ruang', 'count'),
    rata_rata_durasi=('durasi_menit', 'mean'),
    total_pinjam=('jumlah_pinjam', 'sum'),
    rata_rata_kepuasan=('kepuasan', 'mean')
)

# Pembulatan desimal
ringkas['rata_rata_durasi'] = ringkas['rata_rata_durasi'].round(1)
ringkas['rata_rata_kepuasan'] = ringkas['rata_rata_kepuasan'].round(2)

# Urutkan dari rata-rata durasi terbesar
ringkas = ringkas.sort_values(by='rata_rata_durasi', ascending=False)

# Tampilkan
ringkas
```

---
### Tugas 5 (10 poin) — Menyimpan hasil

```python
# Simpan tabel 'ringkas' ke file 'ringkasan_ruang.csv'
ringkas.to_csv('ringkasan_ruang.csv')
```
