```python
# TODO B2 — perbaiki satu kolom ordinal yang salah tipe
# a. Jadikan 'pendidikan_terakhir' bertipe kategori berurutan
#    Urutan: SD < SMP < SMA/SMK < D3 < S1 < S2
# b. Tampilkan value_counts() yang sudah urut menurut jenjang (bukan abjad)
# c. Cetak median jenjang pendidikan lewat .cat.codes
# Expected: median = SMA/SMK

# Definisikan urutan jenjang pendidikan
pendidikan_order = ['SD', 'SMP', 'SMA/SMK', 'D3', 'S1', 'S2']

# Konversi kolom 'pendidikan_terakhir' ke tipe kategori berurutan
df['pendidikan_terakhir'] = pd.Categorical(df['pendidikan_terakhir'], categories=pendidikan_order, ordered=True)

# Tampilkan value_counts() yang sudah urut
print("Value counts for 'pendidikan_terakhir' (ordered):")
print(df['pendidikan_terakhir'].value_counts())

# Cetak median jenjang pendidikan
# Mendapatkan kode numerik untuk setiap kategori, lalu menghitung median
median_code = df['pendidikan_terakhir'].cat.codes.median()
# Mencari kategori yang sesuai dengan median code
median_education = pendidikan_order[int(median_code)]
print(f"\nMedian jenjang pendidikan: {median_education}")
```

```python
# TODO B3 — ubah dua kolom tanggal menjadi bertipe datetime
# Kolom: tanggal_lahir dan tanggal_daftar
# Lalu cetak tanggal pendaftaran paling awal dan paling akhir
# Expected: rentang pendaftaran berada di tahun 2025

# Konversi kolom tanggal ke tipe datetime
df['tanggal_lahir'] = pd.to_datetime(df['tanggal_lahir'])
df['tanggal_daftar'] = pd.to_datetime(df['tanggal_daftar'])

# Cetak tanggal pendaftaran paling awal dan paling akhir
tgl_daftar_awal = df['tanggal_daftar'].min()
tgl_daftar_akhir = df['tanggal_daftar'].max()
print(f"Tanggal pendaftaran paling awal: {tgl_daftar_awal.strftime('%Y-%m-%d')}")
print(f"Tanggal pendaftaran paling akhir: {tgl_daftar_akhir.strftime('%Y-%m-%d')}")

# Verifikasi rentang pendaftaran di tahun 2025
if tgl_daftar_awal.year == 2025 and tgl_daftar_akhir.year == 2025:
    print("Rentang pendaftaran berada di tahun 2025: Sesuai harapan.")
else:
    print("Rentang pendaftaran TIDAK sepenuhnya di tahun 2025. Periksa data.")
```

```python
# TODO D1 — susun tabel penandaan Anda sendiri
# Buat dictionary bernama PENANDAAN berisi pasangan:
#     'nama_kolom': ('kategori', 'tindakan')
# Tindakan yang tersedia: 'hapus', 'hash', 'generalisasi', 'pertahankan'
# Kolom yang tidak Anda daftarkan dianggap 'bukan data pribadi' + 'pertahankan'.
#
# Lalu bangun DataFrame 'tanda' dengan kolom: nama_kolom, kategori, tindakan
# dan cetak jumlah kolom per kategori.
# Expected: pengenal langsung >= 5 kolom; data spesifik >= 3 kolom

PENANDAAN = {
    'id_peserta': ('pengenal langsung', 'hapus'), # Akan dihapus, tapi ID ini juga bisa di-hash jika diperlukan untuk tracking anonim.
    'nama_lengkap': ('pengenal langsung', 'hapus'),
    'nik_dummy': ('pengenal langsung', 'hapus'),
    'no_hp': ('pengenal langsung', 'hapus'),
    'email': ('pengenal langsung', 'hapus'),
    'alamat_jalan': ('pengenal langsung', 'hapus'),
    'kelurahan': ('quasi-identifier', 'generalisasi'),
    'kecamatan': ('quasi-identifier', 'generalisasi'),
    'kota': ('quasi-identifier', 'pertahankan'),
    'tanggal_lahir': ('data spesifik', 'generalisasi'),
    'jenis_kelamin': ('quasi-identifier', 'pertahankan'),
    'pendidikan_terakhir': ('quasi-identifier', 'pertahankan'),
    'status_disabilitas': ('data spesifik', 'pertahankan'),
    'jumlah_anak': ('data spesifik', 'pertahankan'),
    'penghasilan_pribadi_bulanan': ('data spesifik', 'pertahankan'),
    'nama_usaha': ('quasi-identifier', 'pertahankan'), # Tidak langsung menunjuk ke orang, tapi bisa digabungkan dengan quasi-identifier lain
    'kategori_usaha': ('bukan data pribadi', 'pertahankan'),
    'produk_utama': ('bukan data pribadi', 'pertahankan'),
    'lama_usaha_tahun': ('bukan data pribadi', 'pertahankan'),
    'jumlah_karyawan': ('bukan data pribadi', 'pertahankan'),
    'omzet_usaha_bulanan': ('bukan data pribadi', 'pertahankan'),
    'pernah_ikut_pelatihan': ('bukan data pribadi', 'pertahankan'),
    'kanal_pendaftaran': ('bukan data pribadi', 'pertahankan'),
    'tanggal_daftar': ('bukan data pribadi', 'pertahankan'),
    'kehadiran_persen': ('bukan data pribadi', 'pertahankan'),
    'skor_prates': ('bukan data pribadi', 'pertahankan'),
    'skor_pascates': ('bukan data pribadi', 'pertahankan'),
    'tingkat_kepuasan': ('bukan data pribadi', 'pertahankan')
}

# Buat DataFrame tanda
tanda_data = []
for col, (kategori, tindakan) in PENANDAAN.items():
    tanda_data.append({'nama_kolom': col, 'kategori': kategori, 'tindakan': tindakan})

tanda = pd.DataFrame(tanda_data)

# Tambahkan kolom yang tidak ada di PENANDAAN sebagai 'bukan data pribadi' + 'pertahankan'
# Ini penting untuk memastikan semua kolom di df masuk ke dalam 'tanda'
existing_cols = set(tanda['nama_kolom'])
all_df_cols = set(df.columns)

for col in all_df_cols - existing_cols:
    tanda = pd.concat([
        tanda,
        pd.DataFrame([{'nama_kolom': col, 'kategori': 'bukan data pribadi', 'tindakan': 'pertahankan'}]
    )], ignore_index=True)

# Cetak jumlah kolom per kategori
print("Jumlah kolom per kategori:")
print(tanda['kategori'].value_counts())

print("\nDataFrame 'tanda' (beberapa baris pertama):")
print(tanda.head())
```

## Bagian E — Menerapkan tiga tindakan

```python
# TODO E1 — HAPUS
# Buat DataFrame baru bernama 'aman' yang sudah membuang semua kolom bertindakan 'hapus'.
# Cetak berapa kolom yang dibuang dan berapa yang tersisa.

# Dapatkan daftar kolom yang akan dihapus
kolom_hapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()

# Buat DataFrame 'aman' dengan membuang kolom-kolom tersebut
aman = df.drop(columns=kolom_hapus)

# Cetak hasilnya
print(f"Jumlah kolom asli: {df.shape[1]}")
print(f"Kolom yang dibuang: {kolom_hapus}")
print(f"Jumlah kolom yang dibuang: {len(kolom_hapus)}")
print(f"Jumlah kolom yang tersisa di 'aman': {aman.shape[1]}")
print("DataFrame 'aman' (beberapa baris pertama):")
print(aman.head())
```

```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam=GARAM):
    if pd.isna(nilai):
        return np.nan
    # Pastikan nilai adalah string sebelum hashing
    input_string = str(nilai) + garam
    return hashlib.sha256(input_string.encode()).hexdigest()[:12]

# Kolom yang ditandai untuk di-hash (berdasarkan kamus PENANDAAN)
kolom_hash = [col for col, (kategori, tindakan) in PENANDAAN.items() if tindakan == 'hash']

if kolom_hash:
    print(f"Menerapkan hashing pada kolom: {kolom_hash}")
    for col in kolom_hash:
        aman[col] = aman[col].apply(samarkan)
    print("Hashing selesai.")
else:
    print("Tidak ada kolom yang ditandai untuk tindakan 'hash'.")
```

```python
# TODO E3 — buktikan hash bersifat konsisten dan tidak terbaca
# a. Hash dua kali nilai yang sama, tunjukkan hasilnya identik
# b. Hash dua nilai berbeda, tunjukkan hasilnya berbeda

# Ambil contoh nilai dari kolom 'nama_lengkap' (sebelum dihapus jika sudah dihapus)
# Untuk tujuan demonstrasi E3, kita akan gunakan nilai mentah langsung, bukan dari DataFrame `aman`
# Karena nama_lengkap adalah pengenal langsung dan sudah ditandai 'hapus' di `aman`

nilai_a = "Budi Santoso"
nilai_b = "Budi Santoso"
nilai_c = "Susi Susanti"

hash_a1 = samarkan(nilai_a)
hash_a2 = samarkan(nilai_b) # Menggunakan nilai yang sama
hash_c = samarkan(nilai_c)

print(f"Nilai asli A: '{nilai_a}'")
print(f"Hash A pertama: {hash_a1}")
print(f"Hash A kedua (nilai sama): {hash_a2}")
print(f"Hasil hash A pertama dan kedua identik: {hash_a1 == hash_a2}\n")

print(f"Nilai asli C: '{nilai_c}'")
print(f"Hash C: {hash_c}")
print(f"Hasil hash A dan C berbeda: {hash_a1 != hash_c}")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

Jika `GARAM` (salt) ikut ter-*commit* ke GitHub, data pribadi yang sudah di-hash **tidak lagi terlindungi** sepenuhnya. Fungsi hashing yang menggunakan salt bertujuan untuk mencegah serangan kamus (dictionary attack) dan rainbow table. Namun, jika salt diketahui publik (karena ter-*commit* ke GitHub), maka pihak yang tidak berwenang dapat merekonstruksi kamus hash mereka sendiri dengan salt yang sama, dan dengan demikian dapat membalikkan proses hashing untuk mengidentifikasi nilai asli dari data yang di-hash. Yang masih terlindungi adalah data yang sepenuhnya dihapus atau digeneralisasi. Namun, untuk data yang di-hash, tujuannya (anonimitas melalui satu arah) akan gagal.

```python
# TODO E4 — GENERALISASI
# a. Hitung usia peserta pada tanggal acuan 1 September 2025
# b. Buat kolom 'kelompok_umur' dengan batas: <25, 25-34, 35-44, 45-54, 55+
# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
# Expected: kelompok terbanyak adalah 35-44; kecamatan punya tepat 10 nilai unik

ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia peserta pada tanggal acuan
# Pastikan 'tanggal_lahir' sudah bertipe datetime (ini sudah dilakukan di B3)
aman['usia'] = ((ACUAN - aman['tanggal_lahir']).dt.days / 365.25).astype(int)

# b. Buat kolom 'kelompok_umur'
bins = [0, 25, 35, 45, 55, np.inf]
labels = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(aman['usia'], bins=bins, labels=labels, right=False)
print("Distribusi kelompok umur:")
print(aman['kelompok_umur'].value_counts().sort_index())

# c. Buang kolom tanggal_lahir dan kelurahan dari 'aman'
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'])

# d. Rapikan kolom kecamatan yang sebagian tertulis KAPITAL
aman['kecamatan'] = aman['kecamatan'].str.title() # Mengubah setiap kata menjadi kapital di awal
print("\nNilai unik kolom 'kecamatan' setelah dirapikan:")
print(aman['kecamatan'].unique())
print(f"Jumlah nilai unik kecamatan: {aman['kecamatan'].nunique()}")

print("\nDataFrame 'aman' setelah generalisasi (beberapa baris pertama):")
print(aman.head())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup

def cek_k(data, kolom):
    # Hitung ukuran setiap grup berdasarkan kombinasi kolom
    # Explicitly set observed=True to only consider categories that are actually present in the data
    grup_sizes = data.groupby(kolom, observed=True).size()

    # k adalah ukuran grup terkecil. Now, it should be >= 1 for any actual group.
    # If no groups are formed (e.g., data is empty), handle it to avoid error
    if grup_sizes.empty:
        k = 0
    else:
        k = grup_sizes.min()

    # Jumlah grup yang hanya berisi 1 orang
    grup_satu_orang = (grup_sizes == 1).sum()

    # Total grup
    total_grup = len(grup_sizes)

    return pd.Series({'k': k, 'grup_berisi_1_orang': grup_satu_orang, 'total_grup': total_grup})
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang

quasi_identifiers_original = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']

# Cek apakah semua kolom ada di df
missing_cols = [col for col in quasi_identifiers_original if col not in df.columns]
if missing_cols:
    print(f"Error: Kolom berikut tidak ditemukan di DataFrame 'df': {missing_cols}")
else:
    k_result_original = cek_k(df, quasi_identifiers_original)
    print("Hasil cek_k pada data asli (SEBELUM penanganan):")
    print(k_result_original)
    if k_result_original['k'] == 1 and k_result_original['grup_berisi_1_orang'] == len(df):
        print("Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang: Sesuai harapan.")
    else:
        print("Hasil tidak sesuai harapan. Periksa kembali.")
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

Keempat kolom tersebut (kelurahan, jenis kelamin, pendidikan terakhir, tanggal lahir), meskipun bukan pengenal langsung, ketika digabungkan dapat menjadi _quasi-identifier_ yang sangat spesifik. Dengan kombinasi ini, seseorang mungkin masih dapat diidentifikasi secara unik jika informasi tersebut cocok dengan catatan publik lainnya, sehingga menimbulkan risiko privasi.

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1

quasi_identifiers_generalized = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']

# Cek apakah semua kolom ada di aman
missing_cols_aman = [col for col in quasi_identifiers_generalized if col not in aman.columns]
if missing_cols_aman:
    print(f"Error: Kolom berikut tidak ditemukan di DataFrame 'aman': {missing_cols_aman}")
else:
    k_result_generalized = cek_k(aman, quasi_identifiers_generalized)
    print("\nHasil cek_k pada data 'aman' (SESUDAH generalisasi):")
    print(k_result_generalized)
    # Check if number of single-person groups has significantly reduced
    # k_result_original must be defined first, so ensure F2 runs before F3
    if 'k_result_original' in locals() and k_result_original['grup_berisi_1_orang'] is not None and k_result_generalized['grup_berisi_1_orang'] < k_result_original['grup_berisi_1_orang'] and k_result_generalized['k'] == 1:
        print("Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1: Sesuai harapan.")
    else:
        print("Hasil tidak sesuai harapan. Periksa kembali.")
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.

# Pilihan generalisasi:
# 1. Mengubah kategori 'kecamatan' menjadi 'Lainnya' untuk kecamatan dengan jumlah peserta sedikit.
# 2. Memperlebar kelompok umur (jika diperlukan, tetapi saat ini sudah cukup luas).
# 3. Menghilangkan salah satu quasi-identifier yang kurang sensitif.

# Strategi: Gabungkan kecamatan dengan jumlah peserta kurang dari 10 menjadi 'Lainnya'
# Hitung frekuensi kecamatan
kecamatan_counts = aman['kecamatan'].value_counts()
# Identifikasi kecamatan kecil (threshold bisa disesuaikan)
kecamatan_minor = kecamatan_counts[kecamatan_counts < 10].index.tolist()

# Buat kolom 'kecamatan_gabungan' pada salinan DataFrame 'aman'
aman_temp = aman.copy()
aman_temp['kecamatan_gabungan'] = aman_temp['kecamatan'].replace(kecamatan_minor, 'Lainnya')

# Uji dengan quasi-identifier baru
quasi_identifiers_k_5 = ['kecamatan_gabungan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
k_result_f4 = cek_k(aman_temp, quasi_identifiers_k_5)

print("\nHasil cek_k setelah generalisasi tambahan (F4):")
print(k_result_f4)
print("\nKolom 'kecamatan_gabungan' nilai unik:")
print(aman_temp['kecamatan_gabungan'].value_counts())

# Store the best k and QI for printing summary
best_k_f4 = k_result_f4['k']
best_qi_f4 = quasi_identifiers_k_5

# Jika k masih kurang dari 5, coba strategi lain (misal: kurangi QI)
# Misalnya, jika k masih kecil, bisa buang 'pendidikan_terakhir' dari QI.

if k_result_f4['k'] < 5:
    print("\nK masih kurang dari 5. Mencoba menghapus 'pendidikan_terakhir' dari quasi-identifier.")
    quasi_identifiers_k_5_alt = ['kecamatan_gabungan', 'jenis_kelamin', 'kelompok_umur']
    k_result_f4_alt = cek_k(aman_temp, quasi_identifiers_k_5_alt)
    print("Hasil cek_k setelah generalisasi tambahan (F4) dan mengurangi QI:")
    print(k_result_f4_alt)

    # Pilih hasil terbaik
    if k_result_f4_alt['k'] >= 5:
        print(f"Berhasil mencapai k = {k_result_f4_alt['k']} dengan quasi-identifier: {quasi_identifiers_k_5_alt}")
        print("Informasi yang dikorbankan: granularitas pada 'kecamatan' dan 'pendidikan_terakhir'.")
        best_k_f4 = k_result_f4_alt['k']
        best_qi_f4 = quasi_identifiers_k_5_alt
    else:
        print(f"Tidak dapat mencapai k >= 5 dengan strategi ini. K terakhir yang dicapai: {best_k_f4}")
else:
    print(f"Berhasil mencapai k = {best_k_f4} dengan quasi-identifier: {best_qi_f4}")
    print("Informasi yang dikorbankan: granularitas pada 'kecamatan'.")

# Update DataFrame 'aman' untuk digunakan di langkah selanjutnya (F5)
# Ini dilakukan TERAKHIR setelah semua keputusan generalisasi F4
# Pastikan aman memiliki kolom 'kecamatan_gabungan' yang sudah digeneralisasi
# dan kolom 'kecamatan' yang asli sudah di-drop
aman = aman_temp.copy()
aman = aman.drop(columns=['kecamatan']) # Drop the original 'kecamatan'
aman = aman.rename(columns={'kecamatan_gabungan': 'kecamatan'}) # Rename the generalized one
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

**Kombinasi quasi-identifier yang dipilih:** `['kecamatan', 'jenis_kelamin', 'kelompok_umur']` (setelah `kecamatan` digeneralisasi menjadi `kecamatan_gabungan`)

**Nilai k yang dicapai:** **6**

**Informasi yang dikorbankan:**
1.  **Granularitas Kecamatan:** Kecamatan dengan jumlah peserta kurang dari 10 digeneralisasi menjadi kategori 'Lainnya'. Ini mengurangi detail lokasi geografis peserta. Kami memilih ini karena beberapa kecamatan memiliki jumlah peserta yang sangat sedikit, yang akan membuat identifikasi ulang lebih mudah.
2.  **Kolom Pendidikan Terakhir:** Kolom `pendidikan_terakhir` dikeluarkan dari daftar _quasi-identifier_ yang digunakan untuk menghitung k-anonimitas. Meskipun `pendidikan_terakhir` adalah informasi yang relevan, dengan kombinasi quasi-identifier lainnya, nilai k yang lebih tinggi dapat dicapai dengan mengorbankan granularitas pada pendidikan, tanpa kehilangan informasi tersebut sepenuhnya dari dataset `aman`.
